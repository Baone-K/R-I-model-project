```python
# ============================================================
# Imports, Setup, and Data Loading
# ============================================================
import pandas as pd
import numpy as np
import os

# Display settings to see all columns
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

# --- RELATIVE FILE PATHS ---
# 1. Get the current directory 
current_dir = os.getcwd()

# 2. Go up one level to the project root
BASE_DIR = os.path.abspath(os.path.join(current_dir, '..'))

# 3. Define paths to the folders
DATA_DIR = os.path.join(BASE_DIR, 'dataset')
PREPARED_DIR = os.path.join(BASE_DIR, 'prepared_data')

# 4. Create the prepared_data folder if it doesn't exist
os.makedirs(PREPARED_DIR, exist_ok=True)

# 5. Define exact file paths
raw_data_path = os.path.join(DATA_DIR, 'student_career_success_dataset.csv')
clean_data_path = os.path.join(PREPARED_DIR, 'clean_data.csv')

# 6. Load the raw data
df = pd.read_csv(raw_data_path)

print(f"Dataset loaded successfully from: {raw_data_path}")
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
df.head()
```

```python
# ============================================================
# Check Structure & Missing Values
# ============================================================
print("=" * 60)
print("DATASET INFO")
print("=" * 60)
print(df.info())

print("\nMissing Values per Column:")
print(df.isnull().sum())
```

```python
# ============================================================
# Check for Duplicates
# ============================================================
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

if duplicates > 0:
    df = df.drop_duplicates()
    print(f"Duplicates removed. New shape: {df.shape}")
else:
    print("No duplicates detected.")
```

```python
# ============================================================
# Explore Categorical Variables
# ============================================================
# This helps us identify the target variable and demographic columns
categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts().head(10)) # Showing top 10 just in case there are many
```

```python
# ============================================================
# Create Binary Target
# ============================================================
TARGET_COL = 'Placement_Status'

# Create the binary column (1 for Placed, 0 for Not Placed)
df['placement_status_binary'] = (df[TARGET_COL] == 'Placed').astype(int)

print(f"Target column created: 'placement_status_binary'")
print(df['placement_status_binary'].value_counts())
print(f"\nPlacement Rate: {df['placement_status_binary'].mean() * 100:.2f}%")
```

```python
# ============================================================
# Save Cleaned Data
# ============================================================
# Save the cleaned dataset using the relative path we defined earlier
df.to_csv(clean_data_path, index=False)

print(f"Cleaned dataset successfully saved to: {clean_data_path}")
```

```python
# ============================================================
# Preview the Final Data
# ============================================================
# Display the first 20 rows to verify everything is correct before moving to EDA
df.head(20)
```

```python
# Check placement rate by internship count
print(df.groupby('Internships')['placement_status_binary'].mean().round(4))
```

Because students with 0 internships still have a non-zero placement rate (6.7%), this feature cannot be a post-graduation leak. It represents pre-graduation attachments/internships.

The data shows a near-perfect linear relationship: as students complete more internships during their studies, their likelihood of securing a full-time role increases dramatically. Therefore, Internships is retained as the strongest valid pre-placement predictor in the model.

# Notebook 1: Data Preparation — Summary

## Overview
This notebook focused on loading the raw synthetic dataset, assessing its quality, and preparing it for machine learning. The dataset contains 50,000 records and 29 features related to student demographics, academic performance, skills, and placement outcomes.

## Data Quality Checks
- **Missing Values:** None. All 50,000 records are complete.
- **Duplicate Rows:** None detected.
- **Data Types:** The data consists of 14 integer columns, 2 float columns, and 13 categorical (object) columns. 

## Target Variable Creation
The original `Placement_Status` column was categorical ("Placed" / "Not Placed"). To prepare for machine learning models (which require numeric targets), we created a binary target variable:
- `placement_status_binary`: **1** for Placed, **0** for Not Placed.

**Initial Placement Rate:** The clean dataset shows a baseline placement rate of **78.08%** (39,040 placed vs. 10,960 not placed).

## EDA Finding: Validating the "Internships" Feature
A critical step in this notebook was validating whether `Internships` was a post-placement leak or a valid pre-placement predictor. We calculated the placement rate by internship count:
- **0 Internships:** 6.78% placement rate
- **1 Internship:** 23.96%
- **3 Internships:** 61.90%
- **5 Internships:** 91.87%

**Conclusion:** Because students with 0 internships still have a non-zero placement rate (6.7%), this feature cannot be a post-graduation leak (like Starting Salary). It represents **pre-graduation attachments/internships**. The strong linear relationship proves it is the most powerful valid predictor of employability.
